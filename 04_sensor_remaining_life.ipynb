```python
import pandas as pd


X_train = pd.read_csv(
    "17_sensor_replacement_train_X.csv"
)

y_train = pd.read_csv(
    "22_sensor_remaining_life_train_y.csv"
)

X_test = pd.read_csv(
    "19_sensor_replacement_test_X.csv"
)

y_test = pd.read_csv(
    "23_sensor_remaining_life_test_y.csv"
)

y_train = y_train[
    "remaining_useful_life_days"
]

y_test = y_test[
    "remaining_useful_life_days"
]
```

```python
from sklearn.ensemble import RandomForestRegressor


life_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

life_model.fit(
    X_train,
    y_train
)
```

```python
remaining_days = life_model.predict(X_test)
print("Remaining days", remaining_days)
```
